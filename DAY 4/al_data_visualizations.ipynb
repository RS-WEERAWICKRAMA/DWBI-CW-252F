# Sri Lanka A/L Data Analysis (2020)
This notebook analyzes the student dataset and provides visual representations of the performance across different streams and demographics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Upload your CSV file to Google Colab first, or link it to Google Drive.
# df = pd.read_csv('2020_al_data_kaggle_upload_new_old_syllabi.csv')

# If using Google Colab's file upload prompt:
from google.colab import files
uploaded = files.upload()

import io
filename = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[filename]))

# Clean the Zscore column
df['Zscore_numeric'] = pd.to_numeric(df['Zscore'], errors='coerce')
df.head()
```

## 1. Distribution of Students by Stream

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='stream', order=df['stream'].value_counts().index, palette='viridis')
plt.title('Distribution of Students by Stream')
plt.xlabel('Number of Students')
plt.ylabel('Stream')
plt.tight_layout()
plt.show()
```

## 2. Average Z-Score by Stream

```python
plt.figure(figsize=(10, 6))
avg_zscore = df.groupby('stream')['Zscore_numeric'].mean().sort_values()
sns.barplot(x=avg_zscore.values, y=avg_zscore.index, palette='magma')
plt.title('Average Z-Score by Stream')
plt.xlabel('Average Z-Score')
plt.ylabel('Stream')
plt.tight_layout()
plt.show()
```

## 3. Gender Distribution

```python
plt.figure(figsize=(7, 7))
gender_counts = df['gender'].value_counts()
# Filter out malformed data if any
gender_counts = gender_counts[gender_counts.index.isin(['male', 'female'])]
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'], startangle=90)
plt.title('Gender Distribution of Students')
plt.show()
```

## 4. Subject Selection Popularity

```python
plt.figure(figsize=(12, 8))
all_subjects = pd.concat([df['sub1'], df['sub2'], df['sub3']])
top_subjects = all_subjects.value_counts().head(15)
sns.barplot(x=top_subjects.values, y=top_subjects.index, palette='mako')
plt.title('Top 15 Most Popular Subjects')
plt.xlabel('Number of Enrollments')
plt.ylabel('Subject')
plt.tight_layout()
plt.show()
```
